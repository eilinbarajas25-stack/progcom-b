```python
print("=== Simulador de concentracion de un medicamento en la sangre ===")
print("(El cuerpo absorbe el medicamento poco a poco y luego lo va eliminando)\n")

dosis = float(input("Dosis del medicamento en mg: "))
volumen = float(input("Volumen de distribucion en el cuerpo (litros): "))
min_efectivo = float(input("Concentracion minima para que haga efecto (mg/L): "))
max_seguro = float(input("Concentracion maxima segura antes de ser toxica (mg/L): "))

porcentaje_absorcion = 0.25    # cada hora se absorbe el 25% de lo que queda en el estomago
porcentaje_eliminacion = 0.15  # cada hora el cuerpo elimina el 15% de lo que hay en la sangre

dosis_en_estomago = dosis
concentracion = 0.0

historial = {}

for hora in range(0, 25):
    if hora == 0:
        concentracion = 0.0
    else:
        absorbido = dosis_en_estomago * porcentaje_absorcion
        dosis_en_estomago = dosis_en_estomago - absorbido
        concentracion = concentracion + (absorbido / volumen)

        eliminado = concentracion * porcentaje_eliminacion
        concentracion = concentracion - eliminado

    historial[hora] = round(concentracion, 2)

print("\nHora | Concentracion (mg/L) | Nivel")
print("--------------------------------------")

for hora in historial:
    c = historial[hora]
    barras = ""
    for i in range(int(c)):
        barras = barras + "*"
    print(hora, "h  |", c, "mg/L  |", barras)

concentracion_maxima = 0
hora_maxima = 0

for hora in historial:
    if historial[hora] > concentracion_maxima:
        concentracion_maxima = historial[hora]
        hora_maxima = hora

print("\n--- Resultados ---")
print("La concentracion maxima fue de", concentracion_maxima, "mg/L")
print("Se alcanzo en la hora", hora_maxima)

if concentracion_maxima < min_efectivo:
    print("La dosis es INSUFICIENTE, no alcanza a hacer efecto.")
elif concentracion_maxima > max_seguro:
    print("CUIDADO: la dosis es TOXICA, supera el limite seguro.")
else:
    print("La dosis es SEGURA y EFECTIVA.")
```
