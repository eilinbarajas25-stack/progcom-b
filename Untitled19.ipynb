```python
# Piedra, papel, tijera, lagarto, Spock

import random

opciones = ["piedra", "papel", "tijera", "lagarto", "spock"]

# Diccionario: cada opción indica a cuáles le gana
gana_a = {
    "piedra": ["tijera", "lagarto"],
    "papel": ["piedra", "spock"],
    "tijera": ["papel", "lagarto"],
    "lagarto": ["spock", "papel"],
    "spock": ["tijera", "piedra"]
}

print("Opciones disponibles:", opciones)
jugador = input("Elige una opción: ")

maquina = random.choice(opciones)
print("La máquina eligió:", maquina)

if jugador == maquina:
    print("¡Empate!")
elif jugador in gana_a and maquina in gana_a[jugador]:
    print("¡Ganaste!")
elif maquina in gana_a and jugador in gana_a[maquina]:
    print("¡Perdiste!")
else:
    print("Opción no válida")
```
